# Do people regret their vote?

The role of a voter is essential for American democracy, as voters have the power to determine the future of the government. However, sometimes voters are not fully informed. The average American voter is most likely not privy to every candidate's proposed policy, qualifications, and stances on key issues.

Inspired by the 2024 presidential election, I decided to consult Google Trends data to see how many users searched for resources to change their vote, after the election results were announced.

```python
import pandas as pd
import time
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

I used Google Trends to identify how many people looked up "Can I change my vote?" from November 1 to the present day. This time frame will provide an idea of voter regret following the 2024 presidental election. 

```python
vote_df = pd.read_csv('changevote.csv')
```

```python
vote_df = pd.read_csv('changevote.csv',
                      skiprows=2,
                      index_col=0,
                      parse_dates=True)
vote_df = vote_df.replace('<1', 0)
vote_df = vote_df.apply(pd.to_numeric, errors='coerce')
```

```python
ts_plot=vote_df.plot()
plt.show()
```

The graph above shows a strong spike in queries about changing votes shortly after the presidential election results in November 2024, likely driven by widespread public reaction to the outcome. Searches of this nature tend to surge in the immediate aftermath of major elections, as voters process results they may find unexpected or unwelcome.

After that initial spike, query volume dropped significantly and remained relatively low through early 2026.

The spike in April of 2026 can reflect the upcoming midterm elections and growing disapproval for President Trump.

```python
nov_df = vote_df.loc['2024-11']
nov_df.plot(figsize=(12,4))
plt.show()
```

With the largest queries coming from November 2024, I decided to take a closer look at the month and determine what specific weeks had the most searches.

In the first week of November, searches reached around 70, shortly before Election Day, suggesting that some voters were having second thoughts from early voting or mail-in ballots. The search then reached 100 the week of November 10th, the week immediately following the election, after Trump was declared president-elect.

After that, people stopped searching for resources to change their vote after November 17, suggesting that voters accepted the results of the election. 

## Analysis of Trump voters who regret their vote

Considering the high number of people who searched for a resource to change their vote and that Trump was elected president, it implies that people who wanted to change their vote were Trump supporters. With this in mind, I researched to see more demographics about Trump supporters who regret their vote. This information comes from [Navigator's data analysis.](https://navigatorresearch.org/one-in-five-trump-2024-voters-regret-their-vote/)

Navigator found that 1 in 5 Trump supporters regretted their vote after the 2024 election. This statistic reflects the large number of people searching for how to change their vote after the election results were announced.

Navigator also reported that Trump voters that regreted their choice were more likely to be between 18 and 44. However, Trump voters older than 45 were more likely stay confident in their Trump vote.



The graph above suggests that younger voters experienced more regret about voting for Trump. This may be because younger voters are often less politically informed and experienced than older generations. In general, older individuals tend to be more politically engaged, which can make them more likely to feel confident in their voting decisions.

Additionally, people who regreted voting for Trump also expressed a greater disapproval for his policy on tariffs and ICE.

As shown in the graph above, people who regretted voting for Trump are more likely to disapprove of his policies on ICE and tariffs. This sentiment reflects that people who have regret for their vote are more likely to not have been an informed voter and have not been very well-informed on Trump's ICE and tariffs. 

# Conclusion

Overall, the analysis of Google Trends data shows that some voters do regret their vote after election results are announced and the president-elect begins enacting their policies and executive orders.


However, interest in changing votes notable decreases after the second week after the election, implying that voters tend to accept election results and the fact that their vote is final.
